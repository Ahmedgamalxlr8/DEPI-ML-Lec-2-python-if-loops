1.  Get user input using **`input(“Enter your age: ”)`**. If user is 18 or older, give feedback: **`You are old enough to drive`**. If below 18 give feedback to **`You need #number more years to learn to drive`**.
​
Output:
​
```py
    Enter your age: 30
    You are old enough to learn to drive.
    Output:
    Enter your age: 15
    You need 3 more years to learn to drive.
```


````python
age = input("Enter your age: ")
try:
    age = int(age)
except ValueError:
    print("Please enter a valid integer for your age.")
if age >= 18:
    print("You are old enough to drive.")
else:
    print(f"You need {18 - age} more years to learn to drive.")
````


````python
my_age = 21
your_age = input("Enter your age: ")
try: 
    your_age = int(your_age)
except ValueError:
    print("Please enter a valid integer for your age.")
age_diff = abs(my_age - your_age)
if age_diff == 0:
    print("We are the same age.")
elif age_diff == 1:
    if my_age > your_age:
        print("You are a year younger than me.")
    else:
        print("You are a year older than me.")
else:
    if my_age > your_age:
        print(f"You are {age_diff} years younger than me.")
    else:
        print(f"You are {age_diff} years older than me.")  
    
    
````

3. Check if the season is Autumn, Winter, Spring or Summer.
    - If the user input is:
        - September, October or November, the season is Autumn.
        - December, January or February, the season is Winter.
        - March, April or May, the season is Spring
        - June, July or August, the season is Summer


````python
seasons = {"Winter": ["December", "January", "February"],
           "Spring": ["March", "April", "May"],
           "Summer": ["June", "July", "August"],
           "Autumn": ["September", "October", "November"]}

current_month = input("Enter the current month: ")
current_month = current_month.capitalize().strip()  

for season, months in seasons.items():
    if current_month in months:
        print(f"The current season is {season}.")
        break
else:
    print("Invalid month entered.")
````

# Part 2: Dictionary & Conditions Practice


You are given a dictionary called `person` that contains information about an individual.  
Your task is to perform the following operations step-by-step:

##  The Dictionary:

```python
person = {
    'first_name': 'Milaan',
    'last_name': 'Parmar',
    'age': 96,
    'country': 'Finland',
    'is_marred': True,
    'skills': ['Python', 'Matlab', 'R', 'C', 'C++'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}
```

## Tasks

### 1. Check for `skills` key
- If the `skills` key exists in the dictionary:
  - Print the **middle skill** from the list of skills.

### 2. Check for `'Python'` in skills
- Still inside the same condition:
  - Check if `'Python'` is in the person's skills list.
  - Print `True` if found, otherwise `False`.

### 3. Based on the combination of skills:
  - If the person has **only** `['Python', 'Matlab']` → print: `He knows machine learning`
  - If the person has both `'Python'` and `'R'` → print: `He knows statistics`
  - If the person has both `'C'` and `'C++'` → print: `He knows software development`
  - Else → print: `unknown title`


### 4. Print location and marital status
- If the person is married **and** lives in **Finland**, print:
```py
    Milaan Parmar lives in Finland. He is married.
```



````python
person = {
    'first_name': 'Milaan',
    'last_name': 'Parmar',
    'age': 96,
    'country': 'Finland',
    'is_married': True,
    'skills': ['Python', 'Matlab', 'R', 'C', 'C++'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}
````

````python
#1
if 'skills' in person:
    skills = person['skills']
    lenth = len(skills)
    print(f"The middle skill in the skills list is: {skills[lenth // 2]}")
````

````python
#2
if 'skills' in person:
    skills = person['skills']
    if 'Python' in skills:
        print("True")
    else:
        print("False")
````

````python
#3
ml_skills = set(['Python', 'MatLab'])
if set(person['skills']) == ml_skills:
    print("He knows Machine Learning.")
    
if 'Python' in person['skills'] and 'R' in person['skills']:
    print("He knows Statistics.")

if 'C++' in person['skills'] and 'C' in person['skills']:
    print("He knows Software Development.")
else:
    print("Unknown title.")   
````

````python
#4
if person['is_married']:
    print(f"{person['first_name']} {person['last_name']} lives in {person['country']} and is married.")
else:
    print(f"{person['first_name']} {person['last_name']} lives in {person['country']} and is  unmarried.")
````

2. Go to the data folder and use the **[countries_details_data.py]()** file.
   1. What are the total number of languages in the data ( output:  112 )
   2. Find the ten most spoken languages from the data ( output:  English , French , Arabic )
   3. Find the 10 most populated countries in the world ( output: China, India, US )

````python
# 1
from countries_details_dat import countries_data
langs = set()
for country in countries_data:
    langs.update(country['languages'])
langs_num = len(langs)
langs_num
````

````python
languages_countries = {}
for country in countries_data:
    for language in country['languages']:
        if language not in languages_countries:
            languages_countries[language] = []
        languages_countries[language].append(country['name'])
languages_countries[:5]
````

````python
sorted_languages = []
for language, countries in languages_countries.items():
    sorted_languages.append((language, len(countries)))

sorted_languages.sort(key=lambda x: x[1], reverse=True)
top_3_languages = [lang for lang, _ in sorted_languages[:3]]
top_3_languages
````

````python
from countries_details_dat import countries_data

sorted_countries = sorted(
    countries_data,
    key=lambda country: country["population"],
    reverse=True
)
for i, country in enumerate(sorted_countries[:10]):
    print(f"{i + 1}. {country['name']}")
````
